### Problem 1: Explore the Dataset

1. Display the first 10 rows of the dataset.
2. Display 10 random rows.
3. Find the total number of rows and columns.
4. Print the number of rows and the number of columns separately.

```python

import pandas as pd

# Load the dataset
df = pd.read_csv('/content/titanic_data_updated.csv')

# 1. Display the first 10 rows
print("1. First 10 Rows:")
display(df.head(10))

# 2. Display 10 random rows
print("2. 10 Random Rows:")
display(df.sample(n=10, random_state=42))

# 3. Find total number of rows and columns
print("3. Dataset Shape (Rows, Columns):")
print(df.shape)

# 4. Print rows and columns separately
print("\n4. Number of Rows:", df.shape[0])
print("   Number of Columns:", df.shape[1])
```

### Problem 2: Check the Dataset Structure

1. Display the dataset information, including column names, data types, and non-null values.
2. Generate a statistical summary of the numerical columns.
3. Identify which columns contain numerical data and which contain non-numerical data.

```python


# 1. Display dataset information
print("1. Dataset Information:")
df.info()

# 2. Statistical summary of numerical columns
print("\n2. Statistical Summary of Numerical Columns:")
display(df.describe())

# 3. Identify numerical and non-numerical columns
numerical_columns = df.select_dtypes(include=['number']).columns.tolist()
non_numerical_columns = df.select_dtypes(exclude=['number']).columns.tolist()

print("\n3. Numerical Columns:")
print(numerical_columns)

print("\nNon-Numerical Columns:")
print(non_numerical_columns)
```

### Problem 3: Find Missing Values

1. Count the missing values in each column.
2. Display only the columns that contain missing values.
3. Identify the column with the highest number of missing values.

```python


# 1. Count missing values in each column
print("1. Missing Values in Each Column:")
display(df.isnull().sum())

# 2. Display only columns containing missing values
print("\n2. Columns Containing Missing Values:")
missing_columns = df.isnull().sum()
display(missing_columns[missing_columns > 0])

# 3. Identify the column with the highest number of missing values
print("\n3. Column with the Highest Number of Missing Values:")
highest_missing_column = df.isnull().sum().idxmax()
highest_missing_count = df.isnull().sum().max()

print("Column Name:", highest_missing_column)
print("Number of Missing Values:", highest_missing_count)
```

### Problem 4: Check Duplicate Rows

1. Count the total number of duplicate rows.
2. Create a separate DataFrame without duplicate rows.
3. Display the number of rows before and after removing duplicates.

```python


# 1. Count total duplicate rows
duplicate_count = df.duplicated().sum()

print("1. Total Duplicate Rows:", duplicate_count)

# 2. Create a separate DataFrame without duplicate rows
df_cleaned = df.drop_duplicates().copy()

print("\n2. DataFrame After Removing Duplicates:")
display(df_cleaned.head(10))

# 3. Display number of rows before and after removing duplicates
print("\n3. Number of Rows Before and After Removing Duplicates:")
print("Rows Before Removing Duplicates:", df.shape[0])
print("Rows After Removing Duplicates:", df_cleaned.shape[0])
```

### Problem 5: Analyze Passenger Survival

1. Count how many passengers survived and how many did not survive.
2. Calculate the percentage for each group.
3. Create a Seaborn count plot for the `Survived` column.

```python

# 1. Count how many passengers survived and did not survive
survival_count = df['Survived'].value_counts().sort_index()

print("1. Passenger Survival Count:")
print("Did Not Survive (0):", survival_count.get(0, 0))
print("Survived (1):", survival_count.get(1, 0))

# 2. Calculate percentage for each group
survival_percentage = df['Survived'].value_counts(normalize=True).sort_index() * 100

print("\n2. Passenger Survival Percentage:")
print(f"Did Not Survive (0): {survival_percentage.get(0, 0):.2f}%")
print(f"Survived (1): {survival_percentage.get(1, 0):.2f}%")

# 3. Create Seaborn Count Plot
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Survived')

plt.title('Passenger Survival Count')
plt.xlabel('Survival Status (0 = Not Survived, 1 = Survived)')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Not Survived', 'Survived'])

plt.show()
```

### Problem 6: Analyze Categorical Features

For each of these columns — `Pclass`, `Sex`, and `Embarked`:

1. Count the number of records in each category.
2. Calculate the percentage of each category.
3. Create a separate Seaborn count plot.


```python


# Categorical columns
categorical_columns = ['Pclass', 'Sex', 'Embarked']

# Analyze each categorical column
for column in categorical_columns:

    # 1. Count the number of records in each category
    category_count = df[column].value_counts(dropna=False)

    print(f"\n{'='*40}")
    print(f"Analysis of {column}")
    print(f"{'='*40}")

    print("\n1. Category Counts:")
    print(category_count)

    # 2. Calculate percentage of each category
    category_percentage = df[column].value_counts(
        normalize=True, dropna=False
    ) * 100

    print("\n2. Category Percentages:")
    print(category_percentage.round(2))

    # 3. Create a separate Seaborn Count Plot
    plt.figure(figsize=(7, 5))

    sns.countplot(data=df, x=column, order=df[column].value_counts().index)

    plt.title(f'Count Plot of {column}')
    plt.xlabel(column)
    plt.ylabel('Number of Passengers')

    plt.tight_layout()
    plt.show()
```

### Problem 7: Explore the Fare Distribution

1. Create a histogram of the `Fare` column using 50 bins.
2. Create a KDE plot for the `Fare` column.
3. Create a box plot for the `Fare` column.


```python

# 1. Create a Histogram of Fare using 50 bins
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Fare', bins=50, color='skyblue')

plt.title('Distribution of Passenger Fare (Histogram)')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()


# 2. Create a KDE Plot for Fare
plt.figure(figsize=(8, 5))

sns.kdeplot(data=df, x='Fare', fill=True, color='green')

plt.title('KDE Plot of Passenger Fare')
plt.xlabel('Fare')
plt.ylabel('Density')
plt.show()


# 3. Create a Box Plot for Fare
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Fare', color='orange')

plt.title('Box Plot of Passenger Fare')
plt.xlabel('Fare')
plt.show()
```
